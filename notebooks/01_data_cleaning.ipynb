```python
import pandas as pd

# Path to the Excel dataset
file_path = "../data/4877028-Retail_Sales_Data.xlsx"

# Load the Excel file
df = pd.read_excel(file_path, engine="openpyxl")

print("Dataset loaded successfully!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
missing_values = df.isnull().sum()

print("Missing values in each column:")
print(missing_values)
```

```python
duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)
```

```python
df.describe().T
```

```python
# Convert Date column to datetime
df['Date'] = pd.to_datetime(
    df['Date'],
    dayfirst=True,
    errors='coerce'
)

print("Date conversion completed.")
print("Date data type:", df['Date'].dtype)
print("Invalid dates:", df['Date'].isna().sum())
```

```python
print("Minimum Date:", df['Date'].min())
print("Maximum Date:", df['Date'].max())
```

```python
# Clean whitespace from text columns
text_columns = df.select_dtypes(include='object').columns

for column in text_columns:
    df[column] = df[column].str.strip()

print("Text columns cleaned successfully.")
```

```python
print("Negative Unit Price:", (df['Unit_Price'] < 0).sum())
print("Negative Units Sold:", (df['Units_Sold'] < 0).sum())
print("Negative Total Sales:", (df['Total_Sales'] < 0).sum())
print("Negative Discount:", (df['Discount_Percentage'] < 0).sum())
print("Discount above 100%:", (df['Discount_Percentage'] > 100).sum())
print("Negative Revenue:", (df['Revenue'] < 0).sum())
print("Negative Stock:", (df['Stock_On_Hand'] < 0).sum())
```

```python
print("Minimum Store Rating:", df['Store_Rating'].min())
print("Maximum Store Rating:", df['Store_Rating'].max())
```

```python
print("Store Locations:")
print(df['Store_Location'].unique())

print("\nProduct Categories:")
print(df['Product_Category'].unique())

print("\nCustomer Types:")
print(df['Customer_Type'].unique())

print("\nPayment Modes:")
print(df['Payment_Mode'].unique())

print("\nRegions:")
print(df['Region'].unique())

print("\nPromotion Values:")
print(df['Promotion_Applied'].unique())

print("\nHoliday Values:")
print(df['Holiday_Flag'].unique())
```

```python
# Save the cleaned dataset

cleaned_file_path = "../data/cleaned_retail_sales.csv"

df.to_csv(cleaned_file_path, index=False)

print("Cleaned dataset saved successfully!")
print("File:", cleaned_file_path)
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
import pandas as pd

# Load cleaned dataset
cleaned_file_path = "../data/cleaned_retail_sales.csv"

df = pd.read_csv(cleaned_file_path)

# Convert Date back to datetime after reading CSV
df['Date'] = pd.to_datetime(df['Date'])

print("Cleaned dataset loaded!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
# Create date-based features

df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month
df['Month_Name'] = df['Date'].dt.month_name()
df['Quarter'] = df['Date'].dt.quarter
df['Day'] = df['Date'].dt.day
df['Day_of_Week'] = df['Date'].dt.day_name()
df['Week_of_Year'] = df['Date'].dt.isocalendar().week

print("Date features created successfully!")
```
